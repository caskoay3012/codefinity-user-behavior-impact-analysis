## Outlier Detection

## Current Task

Create a function to detect outliers and apply it to both groups.
Create a detect_outliers function that accepts a DataFrame column and:

Normalizes the data using the Box-Cox transformation;

Calculates Q1, Q3, and IQR using np.percentile();

Calculates the lower and upper boundaries (Q1 - 1.5 * IQR, Q3 + 1.5 * IQR);

Returns the indices of outliers.

Apply the function to the following columns and store the results:

data_a["Total_Amount_Spent"] → data_a_spent_outliers;

data_a["AOV"] → data_a_aov_outliers;

data_b["Total_Amount_Spent"] → data_b_spent_outliers;

data_b["AOV"] → data_b_aov_outliers.

```python
# Write your code below
def detect_outliers(column):
    transformed_data, _ = stats.boxcox(column + 1)

    Q1, Q3 = np.percentile(transformed_data, [25, 75])
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return column.index[
        (transformed_data < lower_bound) |
        (transformed_data > upper_bound)
    ]


# Find outliers for each column
data_a_spent_outliers = detect_outliers(data_a["Total_Amount_Spent"])
data_a_aov_outliers = detect_outliers(data_a["AOV"])
data_b_spent_outliers = detect_outliers(data_b["Total_Amount_Spent"])
data_b_aov_outliers = detect_outliers(data_b["AOV"])

# Print counts with clear labels
print(f"Data A — Amount spent: {data_a_spent_outliers.size}")
print(f"Data A — AOV: {data_a_aov_outliers.size}")
print(f"Data B — Amount spent: {data_b_spent_outliers.size}")
print(f"Data B — AOV: {data_b_aov_outliers.size}")      
```

## Formulating Hypotheses & Conducting A/B Test

A/B testing

Now that the dataset is cleaned of outliers, the next step is to test **whether group B performs better than group A**. For this, we use **hypothesis testing** – a statistical method used to determine if a statement about a dataset is likely to be true.

## Hypotheses

In A/B testing, two competing hypotheses are defined:

- **Null hypothesis (H0)**: there is no significant difference between the groups; supported if p-value > 0.05;
- **Alternative hypothesis (H1)**: there is a significant difference; supported if p-value < 0.05.

## ![Formulating Hypotheses](https://content-media-cdn.codefinity.com/courses/b3eef09e-c487-4bf3-814e-9e606e932891/1-12/Formulating+Hypotheses+New.png)

Since the goal is to check if group B **outperforms** group A, directional hypotheses are used:

- **H0**: the values of `Total_Amount_Spent` (and `AOV`) for group B are less than or equal to those for group A;
- **H1**: the values for group B are higher than those for group A.

Note

Rejecting H0 indicates that group B performs better with the new color theme. Failing to reject H0 indicates that group A performs better.
p-value

The **p-value** indicates the likelihood of observing the data if H0 is true:

- a high p-value (> 0.05) means the data is consistent with H0 — do not reject it;
- a low p-value (< 0.05) means the data is unlikely under H0 — reject it.

![Hypothesis Rejection](https://content-media-cdn.codefinity.com/courses/b3eef09e-c487-4bf3-814e-9e606e932891/1-12/Hypothesis+Rejection.png)

## Choosing the Test Method

Since the data follows an **exponential distribution**, the **Mann–Whitney U test** is appropriate. This non-parametric test works for data that does not follow a normal distribution and is available in the `scipy` library.

# Creating a function that performs an A/B test

`def ab_test(sample_a, sample_b, column_name):`
`# Performing the Mann–Whitney U test`
`_, p = stats.mannwhitneyu(`
`sample_a[column_name],`
`sample_b[column_name],`
`alternative="less"`
`)`
`# Interpreting results`
`if p < 0.05:`
`print(f"{column_name}: Group B wins (reject H0)")`
`else:`
`print(f"{column_name}: Group A wins (fail to reject H0)")`
`# Printing the p-value`
`print(f"p-value = {p}")`
`# Running the test for the AOV column`

`ab_test(data_a, data_b, "AOV")`

## Interpreting the Results

- If group B wins (reject H0) – the new color theme should be implemented;
- If group A wins (fail to reject H0) – the original color theme remains, since group B does not show a significant improvement.
